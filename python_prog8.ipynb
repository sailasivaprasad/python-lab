```python
file=open("data.txt","w")
#write
#writelines
content="im a new file"
file.write(content)
print("file created")
file.close()
```

```python
file=open("data.txt","a")
file.writelines(["\n3rd line","\n4th line"])
file.close()
```

```python
with open("data.txt","r") as file:
  content=file.read(10)
  print(content)
  print(file.tell())
  file.seek(2)
  print(file.read())
```

```python
with open("data.txt","r") as file:
  content=file.readlines()
  print(content)
```
