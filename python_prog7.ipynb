```python
#TUPLE OPERATIONS
t1 = (1,2,3)
t2 = (9,7,8)
print(t1 + t2)
print(t1 * 4)
for i in t1:
  print(i)
print(len(t1))
#t1[2] = 12  #not applicable
if "h" in t1:
  print("yes")
else:
  print("no")
```

```python
#TUPLE FUNCTIONS
print(sum(t1))
print(max(t1))
print(min(t1))
```
