```python
# print multiplication table of a given number 2*1=2
n=int(input())
for i in range(2,21):
  print(f"{n} * {i} = {n*i}")
```

```python
# prime number
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("it is not prime")
    break
else:
  print("yes it is prime")
```

```python
# print prime number in between two numbers
a=int(input())
b=int(input())
for i in range(a,b+1):
 for j in range(2,i):
    if(i%j==0):
      break
 else:
  print(i,end=" ")
```

```python
# print the sum of the digits of the given number
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```
